```python
import sys
sys.path.append("/projects/dynaplix/people/zvh378/Cryo-IEF")
import torch
import torch.nn as nn
import torch.nn.functional as F
from safetensors.torch import load_file
import Cryo_IEF.vits as vits

from cryo_sbi.inference.models.embedding_nets import CRYOIEF, CRYOIEF_PATCHED 
```

```python
ckpt_path = "/projects/dynaplix/people/zvh378/cryoBear/CRYO_IEF/cryo_ief_checkpoint/cryo_ief_v1_vit_b/model.safetensors"
save_path = "/projects/dynaplix/people/zvh378/cryoBear/CRYO_IEF/cryo_ief_checkpoint/cryo_ief_v1_vit_b/pretrained_cryoief_patched.pt"
```

```python
def load_embedding(ckpt_path, save_path, dim, patch_size, D: int = 128):
    
    embedding_model = CRYOIEF_PATCHED(dim, D, patch_size)

    # -----------------------------
    # Load pretrained weights
    # -----------------------------

    state_dict = load_file(ckpt_path, device="cpu")

    cleaned_state_dict = {}
    linear_keyword = "head"

    for k, v in state_dict.items():
        if k.startswith("base_encoder") and not k.startswith(f"base_encoder.{linear_keyword}"):
            cleaned_state_dict[k[len("base_encoder."):]] = v
        elif not k.startswith("base_encoder"):
            cleaned_state_dict[k] = v

    # load into model
    embedding_model.load_state_dict(cleaned_state_dict, strict=False)

    # save backbone only
    torch.save(embedding_model.state_dict(), save_path)
    print(f" Pretrained backbone embedding saved: {save_path}")
```

```python
load_embedding(ckpt_path, save_path, dim = 768, patch_size = 14)
```
